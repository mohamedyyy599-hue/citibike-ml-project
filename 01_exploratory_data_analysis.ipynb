```python
import json

with open("data/raw/station_information.json", "r") as f:
    raw_data = json.load(f)

df_stations = pd.json_normalize(raw_data["data"]["stations"])

# Let's compare 'station_id' and 'short_name'
print(df_stations[['station_id', 'short_name', 'name', 'capacity']].head())
```
